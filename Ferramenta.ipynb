```python
def gerar_arquivo_unico(pasta, nome_base="log_dos_testes.txt"):
    """
    Retorna um Path de arquivo que não existe ainda, adicionando número se necessário.
    Exemplo: log_dos_testes.txt -> log_dos_testes_1.txt -> log_dos_testes_2.txt ...
    """
    pasta = Path(pasta)
    arquivo = pasta / nome_base

    if not arquivo.exists():
        return arquivo

    # separa nome e extensão
    nome = arquivo.stem
    ext = arquivo.suffix
    contador = 1

    while True:
        novo_arquivo = pasta / f"{nome}_{contador}{ext}"
        if not novo_arquivo.exists():
            return novo_arquivo
        contador += 1
```

```python
#=====================REGISTROS(LOG)=============================#
NOME_ARQUIVO_LOG = "log_dos_testes.txt"


def registrar_log(mensagem, pasta_logs=None):
    try:
        # caminho padrão Desktop/ferramenta/Logs
        if pasta_logs is None:
            pasta = Path.home() / "Desktop" 
        else:
            pasta = Path(pasta_logs)
        pasta.mkdir(parents=True, exist_ok=True)

        # gera arquivo único
        caminho_completo = gerar_arquivo_unico(pasta, NOME_ARQUIVO_LOG)

        # grava a mensagem
        agora = datetime.now()
        timestamp = agora.strftime("%Y-%m-%d %H:%M:%S")
        mensagem_formatada = f"{timestamp} - {mensagem}\n"

        with open(caminho_completo, 'a', encoding='utf-8') as f:
            f.write(mensagem_formatada)

        print(f"LOG: {mensagem}  (gravado em: {caminho_completo})")

    except Exception as e:
        print(f"ERRO CRÍTICO AO TENTAR REGISTRAR O LOG: {e}")

# Exemplo de uso
registrar_log("Iniciando a função de login...")
registrar_log("Segunda mensagem — novo arquivo se necessário")
```
