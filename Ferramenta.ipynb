```python
#------------------------------DASH----------------------------------#
def Dash(navegador):
  
  # ---- Selecionar Dash ---- #
  rf(x,'//*[@id="navMenuCustom"]/li[1]/a').click(); ts(2)
  rf(x,'//*[@id="select_dashboard"]').click();ts(2)#--> clicando no botão de dashboard
  rf(x,'//*[@id="select_dashboard"]/option[2]').click() #--> opção escolhida 
  time.sleep(2)
  rf(x,'//*[@id="select_dashboard"]/option[1]').click() #--> reverte a opção escolhida 
  time.sleep(2)

  # ---- Avançando/Retraindo dashs ---- #
  rf(x,'//*[@id="previusPageBtn"]').click() #-->Leitura ANTERIOR  de dash
  time.sleep(3)
  rf(x,'//*[@id="nextPageBtn"]').click()# --> PROXIMA leitura da dash
  time.sleep(3)
  rf(x,'//*[@id="getDashInfosNow"]').click(); time.sleep(2) #falta o botao de refresh
  
  # ---- Calendario ---- #
  
  rf(x,f'//*[@id="date_picker_top_dashboard"]').click() ;time.sleep(2)# --> Abre o calendario  
  rf(x,f'/html/body/div[2]/div[1]/div[1]/table/tbody/tr[1]/td[1]').click(); time.sleep(2) # --> Primeiro dia do calendario 

                                      #<MES>#
  rf(x,f'/html/body/div[2]/div[1]/div[1]/table/thead/tr[1]/th[2]/select[1]').click(); time.sleep(2) # --> Botão de lista de mês 
  rf(x,f'/html/body/div[2]/div[1]/div[1]/table/thead/tr[1]/th[2]/select[1]/option[9]').click(); time.sleep(2) # --> Mes (Sentembro) como padrão 

                                      #<ANOS>#  
  rf(x,f'/html/body/div[2]/div[1]/div[1]/table/thead/tr[1]/th[2]/select[2]').click(); time.sleep(2) # --> Botão de lista de Ano
  rf(x,f'/html/body/div[2]/div[1]/div[1]/table/thead/tr[1]/th[2]/select[2]/option[51]').click(); time.sleep(2) # --> Ano (2025) como padrão 
  
                                     #<HORAS>#
  rf(x,f'/html/body/div[2]/div[1]/div[2]/select[1]').click(); time.sleep(2) # --> Botão de lista de Hora
  rf(x,f'/html/body/div[2]/div[1]/div[2]/select[1]/option[1]').click(); time.sleep(2) # --> Hora (1 hora) como padrão

                                    #<MINUTOS>#
  rf(x,f'/html/body/div[2]/div[1]/div[2]/select[2]').click(); time.sleep(2) # --> Botão de lista de Minutos
  rf(x,f'/html/body/div[2]/div[1]/div[2]/select[2]/option[4]').click(); time.sleep(2) # --> Minutos (3 minutos) como padrão

                                    #<PERIODO>#
  rf(x,f'/html/body/div[2]/div[1]/div[2]/select[3]').click(); time.sleep(2) # --> Botão de lista de Periodo            
  rf(x,f'/html/body/div[2]/div[1]/div[2]/select[3]/option[2]').click(); time.sleep(2) # --> Periodo (Pm) como padrão 
    

  rf(x,f'/html/body/div[2]/div[3]/div/button[1]').click(); time.sleep(2) # --> Aplicar
    
  # loop do "cancelar"
  rf(x,'//*[@id="date_picker_top_dashboard"]').click(); time.sleep(2) # --> Abre o calendario 
  rf(x,'/html/body/div[2]/div[3]/div/button[2]').click(); time.sleep(2) # --> Cancelar

  #
  return ("Dash OK")

  
#NOTA : Se execultar varias vezes em seguidas os parametros de data ficam aleatorios :p
    
```

```python
#------------------------------GATEWAY----------------------------------#

def Gateways(navegador, MAC_GW):
  
    rf(x,'//*[@id="navMenuCustom"]/li[5]/a').click();ts(3) #--> Clicando na aba gateways
    rf(x,'//*[@id="datatable_filter_2"]').click();ts(2)#--> Clicando em MAC para pesquisar
    rf(x, '//*[@id="datatable_filter_2"]').send_keys(MAC_GW, Keys.ENTER);ts(3)#--> Clicando na aba de pesquisa e pesquisando oq precisa
    


#---------------------------------------------------------------------------------------------------------------------------------

    rf(x,'//*[@id="tableList"]/tbody/tr[1]/td[1]/a[2]').click();ts(2) #--> Clicando no botão de detalhes
    rf(x,'//*[@id="gateway_operational_status"]').click();ts(2) #--> Clicando no botão de status (ele abre opções)
    rf(x,'//*[@id="gateway_status"]/option[2]').click();ts(2) #--> Clicando no botão da opção 2
    rf(x,'//*[@id="salvar_gateway"]').click();ts(2) #--> Clicando no botão para salva
    


#---------------------------------------------------------------------------------------------------------------------------------

    barra_pesquisa = '//*[@id="tableListResources_filter"]/label/input' #atalho para a aba de pesquisa
    
    rf(x,'//*[@id="btn_resources"]/i').click();ts(2) #--> Clicando no botão recursos
    rf(x,f'{barra_pesquisa}').click();ts(2) #--> Clicando na barra de pesquisa 
    rf(x,f'{barra_pesquisa}').send_keys('txpw');ts(2)  #--> Clicando na aba de pesquisa e pesquisando oq precisa
    rf(x,'//*[@id="tableListResources"]/tbody/tr[1]').click();ts(2) #--> Clicando na opção exata da lista
    rf(x,'//*[@id="btn_get_readings"]').click() #--> Clicando no botão de ler recurso
    time.sleep(5)
    rf(x,'//*[@id="tableListResources_filter"]/label/input').click();ts(2) #--> Clicando na aba de pesquisa
    rf(x,f'{barra_pesquisa}').clear()
    
    rf(x,f'{barra_pesquisa}').send_keys('led');ts(2)  #Clicando na aba de pesquisa e pesquisando oq precisa
    rf(x,'//*[@id="tableListResources"]/tbody/tr').click();ts(2)#--> Clicando na opção exata da lista
    rf(x,'//*[@id="btn_get_readings"]').click();ts(2)#--> Clicando no botão de ler recurso
    rf(x,'//*[@id="tableListResources_filter"]/label/input').click();ts(2) #--> Clicando na barra de pesquisa
    rf(x,f'{barra_pesquisa}').clear()

    rf(x,f'{barra_pesquisa}').send_keys('reset')  #Clicando na aba de pesquisa e pesquisando oq precisa
    time.sleep(2)
    rf(x,'//*[@id="tableListResources"]/tbody/tr[1]').click();ts(2) #--> Clicando na opção exata da lista
    rf(x,'//*[@id="btn_get_readings"]').click();ts(2) #--> Clicando no botão de ler recurso
    rf(x,'//*[@id="resourcesModal"]/div/div/div[1]/button/span').click() #--> Clicando no botão de fechar a aba
    time.sleep(2)


#---------------------------------------------------------------------------------------------------------------------------------
    rf(x,'//*[@id="datatable_filter_2"]').clear()
    rf(x,'//*[@id="datatable_filter_2"]').click();ts(2)#--> Clicando em MAC para pesquisar
    rf(x, '//*[@id="datatable_filter_2"]').send_keys(MAC_GW, Keys.ENTER);ts(3)#--> Clicando na aba de pesquisa e pesquisando oq precisa
    rf(x,'//*[@id="tableList"]/tbody/tr/td[1]/a[1]/span/img').click();ts(2)
    rf(x,'//*[@id="routesUpdate_btn"]').click();ts(2)#--> Clicando no botão de atualizar
    rf(x,'//*[@id="infowindowGW"]/div[1]/div[1]/button[2]').click();ts(5) #--> Clicando no botão de lista Endpoints
    rf(x,'//*[@id="gatewayEPListModal"]/div/div/div[1]/button/span').click();ts(2)
    rf(x,'//*[@id="infowindowGW"]/div[1]/div[1]/button[3]').click() #--> Clicando no botão de Diagrama de rede
    time.sleep(2)
    
    rf(x,'//*[@id="gatewayEPDiagramModal"]/div/div/div[1]/button/span').click();ts(2)
    rf(x,'//*[@id="gw_cmd_select"]').click();ts(2) #--> Clicando no botão de status (ele abre opções)
    rf(x,'//*[@id="gw_cmd_select"]/option[5]').click();ts(2)#--> Clicando no botão da opção 5
    
    rf(x,'//*[@id="gw_cmd_btn"]').click() #--> Clicando no botão enviar
    time.sleep(2)
    
    rf(x,'//*[@id="infoGWModal"]/div/div/div[1]/button/span').click() #--> Clicando no X
    time.sleep(2)

#---------------------------------------------------------------------------------------------------------------------------------

    rf(x,'//*[@id="tableList"]/tbody/tr[1]/td[1]/a[3]/i').click() #--> Clicando no botão de Rotas Ativas
    time.sleep(3)
          
    rf(x,'//*[@id="GWDataModal"]/div/div/div[1]/button/span').click() #--> Clicando no X
    time.sleep(2)
    
    return ("GW ok")
#---------------------------------------------------------------------------------------------------------------------------------
```

```python
#------------------------------AGENDA----------------------------------#

def Agenda(navegador):

    rf(x, '//*[@id="navMenuCustom"]/li[3]/a').click(); ts(5)

    rf(x, '//*[@id="new_item"]').click(); ts(2)
    rf(x, '//*[@id="textNome"]').send_keys('teste')
    rf(x, '//*[@id="textTarefa"]').click()
    rf(x, '//*[@id="textTarefa"]/option[15]').click()
    rf(x, '//*[@id="textRepeticao"]').send_keys('10 minutes')
    rf(x, '//*[@id="salvar_agenda"]').click(); ts(5)

    rf(x, '//*[@id="datatable_filter_1"]').send_keys('teste', Keys.ENTER); ts(2)

    rf(x, '//*[@id="tableList"]/tbody/tr[1]/td[1]').click(); ts(2)
    rf(x, '//*[@id="textNome"]').send_keys('Agenda Teste')
    rf(x, '//*[@id="checkAtivo"]').click()
    rf(x, '//*[@id="salvar_agenda"]').click(); ts(5)

    rf(x, '//*[@id="datatable_filter_1"]').send_keys('Agenda Teste', Keys.ENTER); ts(2)
    rf(x, '//*[@id="tableList"]/tbody/tr[1]/td[1]').click(); ts(2)
    rf(x, '//*[@id="textAreaDados"]').send_keys('teste')
    rf(x, '//*[@id="salvar_agenda"]').click(); ts(1)

    rf(x, '//*[@id="modal_footer"]/button[2]').click(); ts(2)
    rf(x, '//*[@id="tableList"]/tbody/tr[1]/td[1]').click(); ts(2)
    rf(x, '//*[@id="excluir_agenda"]').click(); ts(2)
    navegador.switch_to.alert.accept(); ts(5)

    return ('Página Agenda: OK')
```

```python
def Usuario(navegador):

    # Acessando menu Usuários
    rf(x, '//*[@id="navMenuCustom"]/li[2]/a').click()
    ts(2)

    rf(x, '//*[@id="novo_user"]').click()  # Novo usuário
    ts(2)
    rf(x, '//*[@id="user_name"]').send_keys("Usuario de testes", Keys.ENTER); ts(1)
    rf(x, '//*[@id="user_email"]').send_keys("user1_tests@onerf.com.br", Keys.ENTER); ts(1)
    rf(x, '//*[@id="salvar_user"]').click()  # Salvar
    ts(5)

    # Filtro
    filtro = rf(x, '//*[@id="tableListUsers_filter"]/label/input')
    filtro.send_keys("Usuario de testes", Keys.ENTER)
    ts(1)

    rf(x, '//*[@id="tableListUsers"]/tbody/tr[1]/td[1]/a').click()  # Usuário criado
    ts(1)

    # Editar usuário
    rf(x, '//*[@id="user_name"]').clear()
    rf(x, '//*[@id="user_name"]').send_keys("Usuario de testes - editado", Keys.ENTER)
    rf(x, '//*[@id="user_inactive"]').click(); ts(2)
    rf(x, '//*[@id="salvar_user"]').click(); ts(2)

    # Verificando edição
    rf(x, '//*[@id="navMenuCustom"]/li[2]/a').click()
    rf(x, '//*[@id="tableListUsers_filter"]/label/input').send_keys("Usuario de testes - editado", Keys.ENTER)
    ts(3)

    rf(x, '//*[@id="tableListUsers"]/tbody/tr[1]/td[1]/a').click()  # Selecionando usuário editado
    ts(1)
    rf(x, '//*[@id="excluir_user"]').click()  # Excluir usuário
    navegador.switch_to.alert.accept()  # Confirmar popup
    ts(5)
    
    return ('Usuario OK')
```
